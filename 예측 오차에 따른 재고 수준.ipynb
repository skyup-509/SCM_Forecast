- 판매량을 관측된 수요라고 간주

```python
import pandas as pd

df = pd.read_excel("/content/Online Retail.xlsx")
```

## 1. 일별 SKU별 판매량 정리

1. C로 시작하는 InvoiceNo는 취소 거래로 정의하여 제외
2. 그 외 음수 Quantity는 반품으로 간주하여 정상 판매량에서 차감하여 순판매량(Net Sales) 산출
3. 이후 SKU X 일 단위로 Quantity를 합산하여 순판매량(Net Sales)을 산출한다.

```python
# 1. 취소거래 제외
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]

# 2. SKU x 날짜별 순판매량 계산
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["Date"] = df["InvoiceDate"].dt.date

sku_daily_sales = (
    df.groupby(["StockCode", "Date"], as_index=False)["Quantity"]
      .sum()
      .rename(columns={"Quantity": "NetSales"})
)

# [중요] 거래가 없는 날짜는 순판매량 0
sku_daily_sales["Date"] = pd.to_datetime(sku_daily_sales["Date"])
date_range = pd.date_range(
    sku_daily_sales["Date"].min(),
    sku_daily_sales["Date"].max(),
    freq="D"
)
sku_date = pd.MultiIndex.from_product(
    [sku_daily_sales["StockCode"].unique(), date_range],
    names=["StockCode", "Date"]
).to_frame(index=False)
sku_daily_sales = sku_date.merge(
    sku_daily_sales,
    on=["StockCode", "Date"],
    how="left"
)
sku_daily_sales["NetSales"] = sku_daily_sales["NetSales"].fillna(0)
sku_daily_sales = sku_daily_sales.sort_values(
    ["StockCode", "Date"]
).reset_index(drop=True)
```

```python
sku_daily_sales
```

## 2. 주간 SKU별 판매량 정리

```python
iso_calendar = sku_daily_sales["Date"].dt.isocalendar()

sku_daily_sales["Year"] = iso_calendar.year
sku_daily_sales["Week"] = iso_calendar.week

# 연도-주차 생성
sku_daily_sales["YearWeek"] = (
    sku_daily_sales["Year"].astype(str)
    + "-W"
    + sku_daily_sales["Week"].astype(str).str.zfill(2)
)

# SKU x 연도 x 주차별 순판매량
sku_weekly_sales = (
    sku_daily_sales
    .groupby(["StockCode", "Year", "Week", "YearWeek"], as_index=False)["NetSales"]
    .sum()
)
```

```python
sku_weekly_sales
```

## 3. 가공 데이터가 예측에 용이한 지 판단

### 3-1. 판매량이 0인 주차의 비율이 적었으면 함

```python
sku_summary = (
    sku_weekly_sales
    .groupby("StockCode")["NetSales"]
    .agg(
        mean="mean",
        std="std",
        min="min",
        max="max",
        zero_ratio=lambda x: (x == 0).mean()
    )
    .reset_index()
)

sku_summary.head(10)
```

```python
negative_weekly = sku_weekly_sales[
    sku_weekly_sales["NetSales"] < 0
]

print("음수 주간 판매량:", len(negative_weekly))
print("비율:", len(negative_weekly) / len(sku_weekly_sales))
```

### 3-2. 판매량이 0인 주차의 비율이 전체의 30% 이하, 전체 판매 주차수가 40주차 이상, 평균 판매량이 5 단위 이상

```python
sku_summary = (
    sku_weekly_sales
    .groupby("StockCode")["NetSales"]
    .agg(
        mean="mean",
        std="std",
        zero_ratio=lambda x: (x == 0).mean(),
        weeks="count"
    )
    .reset_index()
)

target_skus = sku_summary[
    (sku_summary["zero_ratio"] <= 0.3) &
    (sku_summary["weeks"] >= 40) &
    (sku_summary["mean"] >= 5)
]["StockCode"]

print("전체 SKU:", sku_summary["StockCode"].nunique())
print("예측 대상 SKU:", target_skus.nunique())

sku_weekly_sales_target = sku_weekly_sales[
    sku_weekly_sales["StockCode"].isin(target_skus)
].copy().reset_index(drop=True)
print(f"후보: {sku_weekly_sales_target['StockCode'].unique()}")
```

## 4. 분석 대상 데이터만 추출

```python
target_skus = [10135, 11001, 15034]

target_data = sku_weekly_sales[
    sku_weekly_sales["StockCode"].isin(target_skus)
].copy().reset_index(drop=True)

target_data.head()
```

```python
target_data.groupby("StockCode")["NetSales"].agg(
    mean="mean",
    std="std",
    zero_ratio=lambda x: (x == 0).mean()
)
```

- zero_ratio는 약 15% 이내로 낮음을 확인
- 표준편차는 60, 혹은 250이므로 상당히 크다는 것을 확인할 수 있음

```python
target_data = target_data.sort_values(
    ["StockCode", "Year", "Week"]
)

target_data.head(20)
```

```python
def add_weekly_features(target_data, df):
    df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

    # ISO 연도 / 주차
    iso_calendar = df["InvoiceDate"].dt.isocalendar()

    df["Year"] = iso_calendar.year
    df["Week"] = iso_calendar.week

    # 3개 SKU에 해당하는 거래만 사용
    target_df = df[
        df["StockCode"].isin(target_data["StockCode"].unique())
    ].copy()

    # SKU × 주별 Feature 생성
    weekly_features = (
        target_df
        .groupby(["StockCode", "Year", "Week"])
        .agg(
            AvgUnitPrice=("UnitPrice", "mean"),
            CustomerCount=("CustomerID", "nunique"),
            TransactionCount=("InvoiceNo", "nunique")
        )
        .reset_index()
    )

    # target_data에 조인
    target_data = target_data.merge(
        weekly_features,
        on=["StockCode", "Year", "Week"],
        how="left"
    )

    return target_data
```

```python
target_data = add_weekly_features(target_data, df)
target_data = target_data.dropna() # AvgUnitPrice, CustomerCount, TransactionCount 연산 결과에 결측이 있는 경우 제거
```

## 5. 파생변수 생성 및 학습/평가 데이터 분할

### 5-1. 파생변수 목록
- 다음 주 수요 (target)
- 과거 판매량 1, 2, 3, 4
- 과거 4주 평균 판매량
- 과거 4주 판매량 표준편차
- 지난 주 AvgUnitPrice, CustomerCount, TransactionCount
- 판매 월

```python
def prepare_forecasting_data(target_data, test_size=0.2):
    target_data = target_data.sort_values(
        ["Year", "Week", "StockCode"]
    ).copy()

    # 다음 주 수요(Target)
    target_data["NextWeekSales"] = (
        target_data
        .groupby("StockCode")["NetSales"]
        .shift(-1)
    )

    # 과거 판매량 Lag
    for lag in [1, 2, 3, 4]:
        target_data[f"lag_{lag}"] = (
            target_data
            .groupby("StockCode")["NetSales"]
            .shift(lag)
        )

    # 과거 4주 평균
    target_data["rolling_mean_4"] = (
        target_data
        .groupby("StockCode")["NetSales"]
        .transform(
            lambda x: x.shift(1).rolling(4).mean()
        )
    )

    # 과거 4주 표준편차
    target_data["rolling_std_4"] = (
        target_data
        .groupby("StockCode")["NetSales"]
        .transform(
            lambda x: x.shift(1).rolling(4).std()
        )
    )

    # 주간 Feature의 직전 주 값
    for col in [
        "AvgUnitPrice",
        "CustomerCount",
        "TransactionCount"
    ]:
        target_data[f"{col}_lag1"] = (
            target_data
            .groupby("StockCode")[col]
            .shift(1)
        )

    # 계절성 Feature
    target_data["Month"] = (
        pd.to_datetime(
            target_data["YearWeek"] + "-1",
            format="%G-W%V-%u"
        ).dt.month
    )

    # 모델에 사용할 Feature
    feature_columns = [
        "StockCode",
        "Week",
        "Month",
        "lag_1",
        "lag_2",
        "lag_3",
        "lag_4",
        "rolling_mean_4",
        "rolling_std_4",
        "AvgUnitPrice_lag1",
        "CustomerCount_lag1",
        "TransactionCount_lag1"
    ]

    target_column = "NextWeekSales"

    # 모델 데이터 생성
    model_data = target_data[
        feature_columns + [target_column]
    ].dropna().copy()

    # StockCode를 범주형 변수로 변환
    model_data["StockCode"] = model_data["StockCode"].astype("category")

    # 시간 순서 기준 정렬
    # Year는 정렬에만 사용하고 모델 Feature에서는 제외
    model_data["_Year"] = target_data.loc[
        model_data.index, "Year"
    ]

    model_data = model_data.sort_values(
        ["_Year", "Week", "StockCode"]
    ).copy()

    # 정렬용 Year 제거
    model_data = model_data.drop(columns=["_Year"])

    # 시간 순서 기준 Train / Test 분리
    split_index = int(
        len(model_data) * (1 - test_size)
    )

    train_data = model_data.iloc[:split_index].copy()
    test_data = model_data.iloc[split_index:].copy()

    X_train = train_data[feature_columns]
    y_train = train_data[target_column]

    X_test = test_data[feature_columns]
    y_test = test_data[target_column]

    return X_train, X_test, y_train, y_test, model_data
```

```python
X_train, X_test, y_train, y_test, model_data = prepare_forecasting_data(target_data)
```

## 6. 예측 모델에 학습 및 예측 결과 산출
### 6-1. CatBoost

```python
from catboost import CatBoostRegressor

cat_model = CatBoostRegressor(
    iterations=300,
    learning_rate=0.05,
    depth=6,
    loss_function="RMSE",
    random_seed=42,
    verbose=False
)

cat_model.fit(
    X_train,
    y_train,
    cat_features=["StockCode"]
)

cat_pred = cat_model.predict(X_test)
```

### 6-2. XGBoost

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    random_state=42,
    n_jobs=-1,
    enable_categorical=True
)

xgb_model.fit(
    X_train,
    y_train
)

xgb_pred = xgb_model.predict(X_test)
```

### 6-3. LightGBM

```python
from lightgbm import LGBMRegressor

lgbm_model = LGBMRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=-1,
    random_state=42,
    n_jobs=-1,
    verbosity=-1
)

lgbm_model.fit(
    X_train,
    y_train,
    categorical_feature=["StockCode"]
)

lgbm_pred = lgbm_model.predict(X_test)
```

## 7. 예측 오차 평가
### 7-1. 예측 데이터 병합

```python
prediction_data = X_test.copy()

prediction_data["Actual"] = y_test.values

prediction_data["CatBoost_Pred"] = cat_pred
prediction_data["XGB_Pred"] = xgb_pred
prediction_data["LGBM_Pred"] = lgbm_pred
```

### 7-2. 예측 오차 부호 계산

```python
prediction_data["CatBoost_Error"] = (
    prediction_data["Actual"]
    - prediction_data["CatBoost_Pred"]
)

prediction_data["XGB_Error"] = (
    prediction_data["Actual"]
    - prediction_data["XGB_Pred"]
)

prediction_data["LGBM_Error"] = (
    prediction_data["Actual"]
    - prediction_data["LGBM_Pred"]
)
```

### 7-3. 모델별 예측 성능 및 오차 계산

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

results = []

for model_name, prediction_column in [
    ("CatBoost", "CatBoost_Pred"),
    ("XGBoost", "XGB_Pred"),
    ("LightGBM", "LGBM_Pred")
]:
    actual = prediction_data["Actual"]
    pred = prediction_data[prediction_column]

    mae = mean_absolute_error(actual, pred)

    rmse = np.sqrt(
        mean_squared_error(actual, pred)
    )

    bias = (actual - pred).mean()

    results.append({
        "Model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "Bias": bias
    })

model_results = pd.DataFrame(results)

model_results
```

## 8. 재고 시뮬레이션
- 목적
  
  수요예측 오차가 실제 재고 운영에 미치는 영향을 비교

- 기본 가정
  + 실제 재고 데이터가 없으므로 가상의 재고 정책을 설정
  + 매주 예측된 다음 주 수요를 기준으로 재고를 준비
  + 안전재고는 고정값으로 설정
- 시뮬레이션 방식
  + 모델별 다음 주 수요 예측
  + 준비 재고 = 예측 수요 + 안전재고
  + 실제 다음 주 판매량과 준비 재고 비교
  + 실제 수요가 준비 재고보다 많으면 품절
  + 준비 재고가 실제 수요보다 많으면 초과재고
- 핵심 산출값 (KPI)
  + 품절 수량
  + 품절 발생 횟수
  + 초과재고 수량
  + 평균 초과재고
- 비교 대상
  + CatBoost
  + XGBoost
  + LightGBM
- 최종 분석
  + 모델별 예측 오차

  $\rightarrow$ Underforecast / Overforecast

  $\rightarrow$ 품절 / 초과재고

  $\rightarrow$ 예측 오차가 재고 운영에 미치는 영향 비교

### 8-1. 안전재고 비율은 0.2로 산정한 품절 및 초과재고 수량

```python
# 안전재고 비율
SAFETY_STOCK_RATE = 0.2

# 모델별 재고 시뮬레이션
for model_name, pred_column in [
    ("CatBoost", "CatBoost_Pred"),
    ("XGBoost", "XGB_Pred"),
    ("LightGBM", "LGBM_Pred")
]:
    # 안전재고
    prediction_data[f"{model_name}_SafetyStock"] = (
        prediction_data[pred_column]
        * SAFETY_STOCK_RATE
    )

    # 준비 재고
    prediction_data[f"{model_name}_Inventory"] = (
        prediction_data[pred_column]
        + prediction_data[f"{model_name}_SafetyStock"]
    )

    # 품절 수량
    prediction_data[f"{model_name}_Stockout"] = (
        prediction_data["Actual"]
        - prediction_data[f"{model_name}_Inventory"]
    ).clip(lower=0)

    # 초과재고 수량
    prediction_data[f"{model_name}_Excess"] = (
        prediction_data[f"{model_name}_Inventory"]
        - prediction_data["Actual"]
    ).clip(lower=0)

```

```python
import matplotlib.pyplot as plt

for model_name, pred_column in [
    ("CatBoost", "CatBoost_Pred"),
    ("XGBoost", "XGB_Pred"),
    ("LightGBM", "LGBM_Pred")
]:

    prediction_data[f"{model_name}_AbsError"] = (
        prediction_data["Actual"]
        - prediction_data[pred_column]
    ).abs()
```

```python
inventory_results = []

for model_name in [
    "CatBoost",
    "XGBoost",
    "LightGBM"
]:
    stockout = prediction_data[
        f"{model_name}_Stockout"
    ]

    excess = prediction_data[
        f"{model_name}_Excess"
    ]

    inventory_results.append({
        "Model": model_name,
        "StockoutQuantity": stockout.sum(),
        "StockoutFrequency": (stockout > 0).mean(),
        "ExcessInventory": excess.sum(),
        "AverageExcessInventory": excess.mean()
    })

inventory_results = pd.DataFrame(inventory_results)

inventory_results
```

```python
plt.figure(figsize=(8, 5))

for model_name in ["CatBoost", "XGBoost", "LightGBM"]:
    plt.scatter(
        prediction_data[f"{model_name}_AbsError"],
        prediction_data[f"{model_name}_Inventory"],
        alpha=0.5,
        label=model_name
    )

plt.xlabel("Absolute Forecast Error")
plt.ylabel("Inventory Level")
plt.title("Forecast Error vs Inventory Level")
plt.legend()
plt.show()
```

### 8-2. 모델별 과대/과소 예측 비율

```python
error_results = []

for model_name, error_column in [
    ("CatBoost", "CatBoost_Error"),
    ("XGBoost", "XGB_Error"),
    ("LightGBM", "LGBM_Error")
]:
    error = prediction_data[error_column]

    error_results.append({
        "Model": model_name,
        "UnderforecastRate": (error > 0).mean(),
        "OverforecastRate": (error < 0).mean(),
        "MeanError": error.mean(),
        "MAE": error.abs().mean()
    })

error_results = pd.DataFrame(error_results)

error_results
```

```python
# 1. 품절 여부 생성
for model_name in ["CatBoost", "XGBoost", "LightGBM"]:
    prediction_data[f"{model_name}_StockoutFlag"] = (
        prediction_data[f"{model_name}_Stockout"] > 0
    ).astype(int)


# 2. 절대 예측 오차 생성
for model_name, pred_column in [
    ("CatBoost", "CatBoost_Pred"),
    ("XGBoost", "XGB_Pred"),
    ("LightGBM", "LGBM_Pred")
]:
    prediction_data[f"{model_name}_AbsError"] = (
        prediction_data["Actual"] - prediction_data[pred_column]
    ).abs()


# 3. 예측 오차를 4개 구간으로 분류
for model_name in ["CatBoost", "XGBoost", "LightGBM"]:
    prediction_data[f"{model_name}_ErrorGroup"] = pd.qcut(
        prediction_data[f"{model_name}_AbsError"],
        q=4,
        duplicates="drop"
    )


# 4. 오차 구간별 품절률 계산
stockout_rate_results = []

for model_name in ["CatBoost", "XGBoost", "LightGBM"]:

    temp = (
        prediction_data
        .groupby(f"{model_name}_ErrorGroup", observed=True)
        .agg(
            StockoutRate=(
                f"{model_name}_StockoutFlag",
                "mean"
            )
        )
        .reset_index()
    )

    temp["Model"] = model_name

    stockout_rate_results.append(temp)


stockout_rate_results = pd.concat(
    stockout_rate_results,
    ignore_index=True
)

stockout_rate_results
```

```python
for model_name in ["CatBoost", "XGBoost", "LightGBM"]:

    temp = stockout_rate_results[
        stockout_rate_results["Model"] == model_name
    ]

    plt.figure(figsize=(8, 5))

    plt.bar(
        temp[f"{model_name}_ErrorGroup"].astype(str),
        temp["StockoutRate"] * 100
    )

    plt.xlabel("Forecast Error Group")
    plt.ylabel("Stockout Rate (%)")
    plt.title(f"{model_name}: Forecast Error vs Stockout Rate")
    plt.xticks(rotation=20)
    plt.show()
```

## 9. 결과 해석

| 모델           | 과소예측 비율 | 과대예측 비율 |    MAE |  품절 수량 |     초과재고 |
| ------------ | ------: | ------: | -----: | -----: | -------: |
| **CatBoost** |   23.1% |   76.9% | 102.57 | 239.96 | 3,051.20 |
| **XGBoost**  |   38.5% |   61.5% |  54.02 | 187.33 | 1,351.99 |
| **LightGBM** |   34.6% |   65.4% |  84.85 | 623.55 | 1,857.24 |

| 예측 오차                    | 재고 영향   |
| ------------------------ | ------- |
| 과소예측 (Actual > Forecast) | 품절 증가   |
